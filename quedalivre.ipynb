```python
import edos
import numpy as np
import matplotlib.pyplot as plt
```

```python
edos.euler
```

```python
fun = lambda t,v: 10-2*v
tempos = np.linspace(0,5,100)
```

```python
valor0 = edos.euler(fun , tempos , 10)
valor1 = edos.euler(fun , tempos , 5)
valor2 = edos.euler(fun , tempos , 0)
valor3 = edos.euler(fun , tempos , -5)
valor4 = edos.euler(fun , tempos , -10)
```

```python
plt.plot(valor0)
plt.plot(valor1)
plt.plot(valor2)
plt.plot(valor3)
plt.plot(valor4)
plt.show()
```
